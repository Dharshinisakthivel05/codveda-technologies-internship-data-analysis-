```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("2) Stock Prices Data Set.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum().sum()
```

```python
numeric_df = df.select_dtypes(include="number")

print("MEAN")
print(numeric_df.mean())

print("\nMEDIAN")
print(numeric_df.median())

print("\nMODE")
print(numeric_df.mode().iloc[0])

print("\nSTANDARD DEVIATION")
print(numeric_df.std())
```

```python
plt.figure(figsize=(8, 5))
df["close"].hist()
plt.title("Distribution of Closing Stock Prices")
plt.xlabel("Closing Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["close"])
plt.title("Boxplot of Closing Stock Prices")
plt.xlabel("Closing Price")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["open"], df["close"])
plt.title("Opening Price vs Closing Price")
plt.xlabel("Opening Price")
plt.ylabel("Closing Price")
plt.show()
```

```python
corr = numeric_df.corr()
print(corr)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Stock Price Correlation Heatmap")
plt.show()
```
